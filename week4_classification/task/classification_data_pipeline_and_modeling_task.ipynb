## Dataset Overview: Loan Approval Classification
**MCC Data Science Track — Week 4: Classification (Task)**

This dataset contains **45,000 records** and **14 columns** (13 predictive features and 1 binary target) designed for binary classification modeling. The goal is to predict whether a loan application will be **approved (`1`)** or **rejected (`0`)** based on the applicant's demographics, financial background, loan details, and credit history.

### Features Description
| Feature | Data Type | Role | Description |
| :--- | :--- | :--- | :--- |
| **`person_age`** | Numerical | Feature | Age of the applicant (years) |
| **`person_gender`** | Categorical | Feature | Gender of the applicant (`male`, `female`) |
| **`person_education`** | Categorical (Ordinal) | Feature | Highest education level (`High School`, `Associate`, `Bachelor`, `Master`, `Doctorate`) |
| **`person_income`** | Numerical | Feature | Annual income of the applicant |
| **`person_emp_exp`** | Numerical | Feature | Years of employment experience |
| **`person_home_ownership`** | Categorical | Feature | Home ownership status (`RENT`, `MORTGAGE`, `OWN`, `OTHER`) |
| **`loan_amnt`** | Numerical | Feature | Loan amount requested |
| **`loan_intent`** | Categorical | Feature | Purpose of the loan (`EDUCATION`, `MEDICAL`, `VENTURE`, `PERSONAL`, `DEBTCONSOLIDATION`, `HOMEIMPROVEMENT`) |
| **`loan_int_rate`** | Numerical | Feature | Loan interest rate (%) |
| **`loan_percent_income`** | Numerical | Feature | Loan amount as a proportion of annual income |
| **`cb_person_cred_hist_length`** | Numerical | Feature | Length of the applicant's credit history (years) |
| **`credit_score`** | Numerical | Feature | Applicant's credit score |
| **`previous_loan_defaults_on_file`** | Categorical | Feature | Whether the applicant has previous loan defaults on record (`Yes`, `No`) |
| **`loan_status`** | Numerical (Binary) | **Target ($y$)** | Loan approval decision (`1`: Approved, `0`: Rejected) |

**Source:** [Kaggle — Loan Approval Classification Dataset](https://www.kaggle.com/datasets/taweilo/loan-approval-classification-data)

# 3. Preprocessing

1. **Feature and Target Separation**: Separate the feature matrix ($X$) and the target vector ($y$ — `loan_status`).
2. **Train-Test Splitting**: Split the dataset into training ($80\%$) and testing ($20\%$) sets using `train_test_split`.
   - Ensure you set `random_state=42` for reproducibility.
   - Use `stratify=y` to preserve the original class proportion ($78\%$ rejected vs. $22\%$ approved) across both splits.
3. **Categorical Encoding**:
   - Construct a `ColumnTransformer` to handle both nominal and ordinal categorical features:
     - **Ordinal Encoding**: Apply `OrdinalEncoder` to `person_education` using the natural hierarchical order: `['High School', 'Associate', 'Bachelor', 'Master', 'Doctorate']`.
     - **Nominal Encoding**: Apply `OneHotEncoder(drop='if_binary', sparse_output=False, handle_unknown='ignore')` to nominal and binary features (`person_gender`, `person_home_ownership`, `loan_intent`, `previous_loan_defaults_on_file`).
     - Set `remainder='passthrough'` and configure `verbose_feature_names_out=False` to retain clean column names.
4. **Feature Scaling**:
   - Apply `StandardScaler` to scale all continuous numerical features.
   - Fit the scaler **strictly** on the training data (`X_train`), then transform both the training data (`X_train`) and testing data (`X_test`).
5. **Target Verification**: Verify that the target vectors (`y_train`, `y_test`) are formatted appropriately as binary integer arrays.

> **Hints:**
> - **Data Leakage**: Never fit scalers or encoders on the full dataset before splitting! All transformations must learn parameters (mean, standard deviation, category mappings) solely from the training set ($X_{train}$) and apply them to the test set ($X_{test}$).
> - **Class Imbalance**: Keep in mind that loan approval is an imbalanced classification problem (~78% class 0 vs ~22% class 1). Accuracy alone can be misleading; precision, recall, and F1-score are critical.
